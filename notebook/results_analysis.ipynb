# Multi-LLM Collaborative Debate System — Results Analysis

This notebook analyzes the results of the Multi-LLM Collaborative Debate System.

The experiment evaluates:
- Single-LLM baseline
- Majority-vote baseline
- Full debate system
- Solver refinement
- Improvement rate
- Consensus rate

## 1. Experimental Setup

The evaluation was performed on 25 challenging reasoning problems.

For each question, four LLM agents were used:
- 3 Solver agents
- 1 Judge agent

The system consisted of the following stages:

1. Role self-assessment
2. Deterministic role assignment
3. Independent solving
4. Peer review
5. Solution refinement
6. Final judging

The experiment also included two baselines:
- Single-LLM baseline
- Majority-vote baseline

## 2. Main Results

| System | Accuracy |
|---|---:|
| Single LLM | 28.00% |
| Initial Solvers | 41.33% |
| Majority Vote | 48.00% |
| Full Debate / Judge | 48.00% |
| Refined Solvers | 38.67% |

The Single-LLM baseline achieved 28% accuracy.

Majority voting increased accuracy to 48%.

The Full Debate system also achieved 48% accuracy, so it did not outperform the simpler majority-vote baseline in this experiment.

```python
from IPython.display import Image, display

display(Image("../analysis/plots/accuracy_comparison.png"))
```

## 3. Solver Refinement

The initial Solver accuracy was 41.33%.

After peer review and refinement, Solver accuracy decreased to 38.67%.

The Improvement Rate was 0%, meaning that none of the initially incorrect Solver answers became correct after refinement.

The experiment also showed that some initially correct answers became incorrect after refinement.

This demonstrates that peer feedback is not always reliable and that an LLM may sometimes accept an incorrect criticism.

```python
display(Image("../analysis/plots/solver_refinement.png"))
```

## 4. Improvement and Degradation

The refinement process was evaluated in two directions:

- Improvement: Wrong → Correct
- Degradation: Correct → Wrong

The Improvement Rate was 0%.

The Degradation Rate measures how often refinement changed an initially correct answer into an incorrect answer.

This is important because refinement does not automatically improve an LLM's reasoning.

display(Image("../analysis/plots/improvement_vs_degradation.png"))

## 5. Consensus

The three refined Solvers produced the same answer on 7 out of 25 questions.

Therefore:

Consensus Rate = 7 / 25 = 28%

This indicates that the Solvers frequently disagreed with one another.

display(Image("../analysis/plots/consensus_rate.png"))

## 6. Comparison With Baselines

The Single-LLM baseline achieved 28% accuracy.

Majority voting achieved 48% accuracy, which was 20 percentage points higher than the Single-LLM baseline.

The Full Debate system also achieved 48% accuracy.

Therefore, in this experiment, the additional peer-review and refinement stages did not produce higher final accuracy than simple majority voting.

## 7. Limitations

Several limitations should be considered when interpreting the results.

1. The benchmark contained only 25 questions.
2. All four agents used the same underlying LLM model.
3. LLM outputs can vary between API calls.
4. Peer reviews can contain incorrect criticism.
5. The refinement stage can sometimes cause correct solutions to become incorrect.
6. The results therefore describe this particular model, prompt design, and benchmark rather than LLM debate systems in general.

## 8. Conclusion

The experiment showed that multiple independent solutions combined through majority voting performed better than the Single-LLM baseline.

However, the Full Debate system achieved the same 48% accuracy as majority voting.

The refinement stage reduced individual Solver accuracy from 41.33% to 38.67%, demonstrating that peer feedback can sometimes introduce errors instead of correcting them.

These results show that adding more collaborative stages does not necessarily guarantee better accuracy. The reliability of peer review and refinement is an important factor in multi-LLM collaborative reasoning systems.
